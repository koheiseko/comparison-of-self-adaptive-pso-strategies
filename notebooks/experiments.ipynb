# Execução dos experimentos

## Protocolo experimental

| Parâmetro | Configuração |
|---|---:|
| Execuções independentes | 20 por cenário |
| Dimensões | 30, 50 e 100 |
| Iterações em 30D | 6.000 |
| Iterações em 50D | 10.000 |
| Iterações em 100D | 20.000 |
| Processos paralelos | 8 |
| Semente base | 42 |
| Medição de memória | Habilitada |

## Setup

```python
from configs.benchmarks import BenchmarksCEC2017Config
from configs.hyperparameters import Algorithms, HyperparametersConfig
from scripts.run_experiments import ExperimentsConfig, run_experiments
from src.pso import APSO, APSOVI, PSOLVIW, PSOTVAC, UAPSO, PSOVectorized
```

## 1. Configuração 

```python
algorithms_config = Algorithms(
    PSO=PSOVectorized,
    PSOLVIW=PSOLVIW,
    PSOTVAC=PSOTVAC,
    APSOVI=APSOVI,
    APSO=APSO,
    UAPSO=UAPSO,
)

experiment_config = ExperimentsConfig(
    algorithms=algorithms_config,
    config_hyperparameters=HyperparametersConfig(),
    config_benchmarks=BenchmarksCEC2017Config(),
    output_dir="../results",
    n_runs=20,
    dim=[30, 50, 100],
    n_iterations_by_dim={
        30: 6_000,
        50: 10_000,
        100: 20_000,
    },
    n_jobs=8,
    seed=42,
    measure_memory=True,
    fail_on_error=True,
)
```

## 2. Execução


```python
run_experiments(experiment_config)
```
